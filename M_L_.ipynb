```python
n = 10
total = 0

for i in range(1, n + 1):
    total += i

formula_result = n * (n + 1) // 2

print(f'Sum (loop)    : {total}')
print(f'Sum (formula) : {formula_result}')
```

```python
def square(x):
    """Returns the square of x."""
    result = x ** 2
    return result


print("===== Square Function =====")
print(square(4))      # 16
print(square(9))      # 81
print(square(-3))     # 9

```

```python
import math


def circle_area(r):
    """Returns area of a circle: A = πr²"""
    return math.pi * r ** 2


def circle_circumference(r):
    """Returns circumference: C = 2πr"""
    return 2 * math.pi * r


def triangle_area(base, height):
    """Returns area of a triangle."""
    return 0.5 * base * height


print("\n===== Geometry Functions =====")

r = 5

print(f"Circle Area (r={r})         : {circle_area(r):.4f}")
print(f"Circle Circumference (r={r}): {circle_circumference(r):.4f}")
print(f"Triangle Area (6,4)         : {triangle_area(6,4):.4f}")
```

```python
def power(base, exponent=2):
    """Raises base to exponent. Default exponent = 2."""
    return base ** exponent


print("\n===== Default Parameters =====")

print(power(3))
print(power(3, 3))
print(power(2, 10))
```

```python
def quadratic_roots(a, b, c):
    """
    Returns the two real roots of
    ax² + bx + c = 0.
    """

    discriminant = b**2 - 4*a*c

    if discriminant < 0:
        return None, None

    root1 = (-b + discriminant**0.5) / (2*a)
    root2 = (-b - discriminant**0.5) / (2*a)

    return root1, root2


print("\n===== Quadratic Formula =====")

r1, r2 = quadratic_roots(1, -5, 6)
print(f"x² - 5x + 6 = 0")
print(f"Root1 = {r1}")
print(f"Root2 = {r2}")

print()

r1, r2 = quadratic_roots(1, 0, 1)
print("x² + 1 = 0")
print(f"Root1 = {r1}")
print(f"Root2 = {r2}")
```

```python
double = lambda x: x * 2
cube = lambda x: x ** 3
celsius_to_fahrenheit = lambda c: (c * 9/5) + 32


print("\n===== Lambda Functions =====")

print(double(7))
print(cube(4))
print(celsius_to_fahrenheit(100))
```

```python

score = 73

if score >= 90:
    grade = "A (Excellent)"
elif score >= 80:
    grade = "B (Good)"
elif score >= 70:
    grade = "C (Average)"
elif score >= 60:
    grade = "D (Below Average)"
else:
    grade = "F (Fail)"

print(f"Score: {score} → Grade: {grade}")
```

```python

x = 15

# AND
if x > 10 and x < 20:
    print(f"{x} is between 10 and 20")

# OR
if x < 5 or x > 12:
    print(f"{x} is outside the range 5–12")

# NOT
if not x == 10:
    print(f"{x} is not equal to 10")
```

```python

n = 36

# Even or Odd
if n % 2 == 0:
    print(f"{n} is Even")
else:
    print(f"{n} is Odd")

# Divisibility by 3 and 5
if n % 3 == 0 and n % 5 == 0:
    print(f"{n} is divisible by both 3 and 5")
elif n % 3 == 0:
    print(f"{n} is divisible by 3 only")
elif n % 5 == 0:
    print(f"{n} is divisible by 5 only")
```

```python


import math

deg = 45
rad = math.radians(deg)

print(f"Angle: {deg}° = {rad:.6f} radians")
print(f"sin(45°) = {math.sin(rad):.6f}")
print(f"cos(45°) = {math.cos(rad):.6f}")
print(f"tan(45°) = {math.tan(rad):.6f}")

# Verify Pythagorean Identity
identity = math.sin(rad)**2 + math.cos(rad)**2
print(f"sin²(45°) + cos²(45°) = {identity:.6f}")
```
